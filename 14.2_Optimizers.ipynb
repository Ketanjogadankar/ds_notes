# Optimizers

Optimization algorithms help train any neural network and modify the model parameters to minimize loss i.e. they help find optimal value for parameter where loss would be minimum.

## Gradient Descent or Standard Gradient Descent

Standard Gradient descent or Batch gradient descent which uses gradients or slop of the loss function w.r.t. corresponding parameter to help find global minima so that loss would be minimum.<br>

It uses complete dataset in a single epoch to take step towards minima.
because of this it needs higher memory and its quite slow. There are possibilities of it getting stuck into local minima.<br>

In Standard gradient descent, **all parameters share same learning rate.**

To over-come this we have mini-batch gradient descent or stochastic gradient descent.

$$
\theta = \theta - \alpha * g_{t}
$$

where $\alpha$ is learning rate, $\theta$ is the model parameter and $g_{t}$ is a gradient of the cost function w.r.t model paramater.

**This update changes the parameters θ in the negative direction of the gradient to minimize the cost function. The learning rate α determines the size of the step.**

## Mini-Batch Gradient Descent

In mini-batch gradient descent batch size is very small per epoch. Normally it has batch size of 32,64,128,256,512.
Its a faster approach and requires less memory. Works well with larger datasets.
Smoother than Stochastic gradient descent.

## Stochastic Gradient Descent

If batch size is equal to 1 then its called a stochastic gradient descent.
SGD is faster, requires small memory, can escape local minima, better at handling larger datasets but noisy and unstable.

## Momentum
For faster convergence, momentum helps speeding or accelerating gradients in the right direction by adding a fraction of the previous gradient to the current one.
<br>

For example, let’s say a gradient has been consistently pointing in the same direction. The momentum term proportional to the previous gradients will accumulate and accelerate the optimization in that direction.

Also it helps smooth out the gradient descent without being stuck in the local minima.
Mathematical representation of same is:

$$
v_t = \beta \cdot v_{t-1} + (1 - \beta) \cdot \nabla J(\theta) \\
\theta = \theta - \alpha \cdot v_t
$$

where, <br>
- $v_t$ is the velocity,<br>
- $\beta$ is the momentum coefficient, typically 0.9<br>
- $\nabla J(\theta)$ is current gradient<br>
- $\alpha$ is learning rate

## RMSProp (Root Mean Square Propagation)

RMSProp is an adaptive learning rate optimization algorithm — it adjusts the learning rate for each parameter individually based on the recent magnitude of its gradients.<br>

If a parameter’s gradient is consistently large, RMSProp will reduce the learning rate for that parameter i.e. will get smaller update steps.<br>
If low gradients then it will increase the learning rate i.e. bigger steps.<br> 
So, while Momentum is about accelerating in consistent directions, RMSProp is based on controlling overshooting by modulating step size. 

Mathematical representation
$$
E[g^2]_t = \gamma \cdot E[g^2]_{t-1} + (1 - \gamma) \cdot g_t^2 \\
\theta_{t+1} = \theta_t - \frac{\alpha}{\sqrt{E[g^2]_t} + \epsilon} \cdot g_t
$$

## Adam (Adaptive Moment Estimation)

Adam combines the benefits of two other extensions of stochastic gradient descent:

Momentum – which helps smooth gradients by considering past gradients (like a moving average) and accelerate gradient descent.

RMSProp – which adapts the learning rate for each parameter based on the magnitude of recent gradients. Also it modulates the step size as per gradient value.

It maintains two moving averages for each parameter:

𝑚_𝑡: First moment (mean of gradients)<br>
v_t: Second moment (uncentered variance of gradients)

- First moment estimate (momentum)
$$
m_t = \beta_1 \cdot m_{t-1} + (1 - \beta_1) \cdot g_t
$$
- Second moment estimate (squared gradients)
$$
v_t = \beta_2 \cdot v_{t-1} + (1 - \beta_2) \cdot g_t^2
$$

- Bias-corrected first and second moments
$$
\hat{m}_t = \frac{m_t}{1 - \beta_1^t}, \quad \hat{v}_t = \frac{v_t}{1 - \beta_2^t}
$$

- Parameter update rule
$$
\theta_{t+1} = \theta_t - \alpha \cdot \frac{\hat{m}_t}{\sqrt{\hat{v}_t} + \epsilon}
$$

Where,
-  $\alpha$: Learning rate (default: 0.001)
- $\beta_1$: Exponential decay rate for the first moment estimate (default: 0.9)
- $\beta_2$: Exponential decay rate for the second moment estimate (default: 0.999)
- $\epsilon$: Small constant for numerical stability (default: $10^{-8}$)
